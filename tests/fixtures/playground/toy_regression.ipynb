# Fitting a line by gradient descent

A synthetic exercise: recover the slope of noisy points.
<img src="https://example.org/diagram.png" width=400>


```python
import numpy as np
import pandas as pd
# see https://numpy.org/doc/ for the API
```

```python
points = pd.read_csv("https://example.org/data/points.csv")
# extra = pd.read_csv('https://example.org/data/extra.csv')
points.head()
```

## Gradient descent
Update the slope a little at a time.


```python
w = 0.0
lr = 0.01
for step in range(100):
    grad = -2 * np.mean(points.x * (points.y - w * points.x))
    w -= lr * grad
print(f"slope={w:.3f}")
```

```python
import matplotlib.pyplot as plt
plt.scatter(points.x, points.y)
plt.show()
```
